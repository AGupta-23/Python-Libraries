## Object Oriented Way

To select the canvas

fig.add_axes() is used to manually place graphs inside the figure.

Syntax:

[left, bottom, width, height]

Values must be between 0 and 1.

```python
fig=plt.figure()
# just a canvas right now
axis1=fig.add_axes([0.1,0.1,1,1]) # type: ignore
# [left se how much % shift, bottom se how much,80% will be width, 80% height
#values much be between 0 and 1
axis1.plot(x,y)

axis2=fig.add_axes([0.3,0.4,0.3,0.3]) #type: ignore
axis2.plot(x,y)
```

Matplotlib allows the aspect ratio, DPI and figure size to be specified when the Figure object is created. You can use the figsize and dpi keyword arguments.

figsize is a tuple of the width and height of the figure in inches dpi is the dots-per-inch (pixel per inch).

```python
a = np.arange(0,10)
b = a * 2

# figsize = (width,height)
# dpi = image quality

fig = plt.figure(figsize=(6,4), dpi=300)

# Main graph
axis1 = fig.add_axes([0.1,0.1,0.8,0.8]) # type: ignore
axis1.plot(a,b)

# Small graph inside main graph
axis2 = fig.add_axes([0.6,0.5,0.3,0.3]) # type: ignore
axis2.plot(a,b)

plt.show()
```
