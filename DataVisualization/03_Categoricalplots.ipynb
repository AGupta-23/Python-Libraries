# Categorical Data Plots

Now let's discuss using seaborn to plot categorical data! There are a few main plot types for this:

* boxplot
* violinplot
* stripplot
* swarmplot
* barplot
* countplot

Let's go through examples of each!

```python
df = sns.load_dataset("tips")
df
```

```python
sns.countplot(x = df['sex'],hue = df['smoker'])
```

```python
sns.barplot(x = df['sex'],y = df['total_bill'])
#takes 2 positionals x and y
# y - quantitative and x= qualitative
```

You can change the estimator object to your own function, that converts a vector to a scalar
- Mean (default): np.mean
- Median: np.median
- Sum: np.sum
- Count: len (to count the number of observations)
- Standard Deviation: np.std
- Minimum: np.min
- Maximum: np.max

```python
import numpy as np 
sns.barplot(x = df['sex'],y = df['tip'],estimator=np.sum)
# y pe tip - sum func
```

```python
sns.boxplot(x= 'tip',y = 'day',data = df,palette = 'rainbow')
# x - quantitative y- qualitative 
# dots - outliers
```

## boxplot and violinplot

boxplots and violinplots are used to shown the distribution of categorical data. A box plot (or box-and-whisker plot) shows the distribution of quantitative data in a way that facilitates comparisons between variables or across levels of a categorical variable. The box shows the quartiles of the dataset while the whiskers extend to show the rest of the distribution, except for points that are determined to be “outliers” using a method that is a function of the inter-quartile range.

```python
sns.violinplot(x = 'day',y = 'total_bill',data = df)
#shows data variation - shows spread etc
```

## stripplot and swarmplot
The stripplot will draw a scatterplot where one variable is categorical. A strip plot can be drawn on its own, but it is also a good complement to a box or violin plot in cases where you want to show all observations along with some representation of the underlying distribution.

The swarmplot is similar to stripplot(), but the points are adjusted (only along the categorical axis) so that they don’t overlap. This gives a better representation of the distribution of values, although it does not scale as well to large numbers of observations (both in terms of the ability to show all the points and in terms of the computation needed to arrange them).

```python
sns.stripplot(x = 'day',y = 'total_bill',data = df)
```

```python
sns.swarmplot(x = 'day',y = 'total_bill',data = df)
#works more spreaded like violin plot
```

```python
sns.violinplot(x= 'tip',y = 'day',data = df)
sns.swarmplot(x= 'tip',y = 'day',data = df)
```

# Categorical Plots

Categorical plots are used to analyze and visualize relationships between:

* categorical data
* numerical data
* category frequencies

These plots are very important in:

* Exploratory Data Analysis (EDA)
* Business analytics
* Data Science

They help compare categories, distributions, counts, and patterns.

---

# 1. Boxplot

A boxplot is used to visualize the statistical distribution of numerical data across categories.

It shows:

* minimum value
* first quartile (Q1)
* median
* third quartile (Q3)
* maximum value
* outliers

### Main Purpose

* Compare distributions between categories
* Detect outliers
* Understand spread and variability

### Example Use Cases

* Salary distribution department-wise
* Marks distribution class-wise

### Advantages

* Very compact statistical summary
* Excellent for outlier detection

---

# 2. Violinplot

A violinplot is an advanced version of a boxplot that also displays probability density.

It combines:

* boxplot
* KDE (Kernel Density Estimation)

### Main Purpose

* Understand data distribution shape
* Compare category-wise densities

### Example Use Cases

* Distribution of customer spending by region
* Age distribution by gender

### Advantages

* More detailed than boxplot
* Shows concentration of values

---

# 3. Stripplot

A stripplot displays individual data points for categorical data.

Each point represents:

* one observation

### Main Purpose

* Visualize actual observations
* Understand spread and clustering

### Example Use Cases

* Individual student marks by section
* Product prices by category

### Characteristics

* Simple and lightweight
* Points may overlap if data is dense

---

# 4. Swarmplot

A swarmplot is similar to a stripplot but automatically adjusts points to prevent overlap.

### Main Purpose

* Clearly visualize all observations
* Avoid overlapping points

### Example Use Cases

* Visualizing dense categorical datasets
* Comparing distributions clearly

### Advantages

* More readable than stripplot
* Better for medium-sized datasets

---

# 5. Barplot

A barplot shows aggregated numerical values for different categories using bars.

Usually represents:

* mean
* average
* sum
* statistical aggregations

### Main Purpose

* Compare category-wise numerical metrics

### Example Use Cases

* Average sales by region
* Average salary by department

### Important Note

Barplot does NOT show raw counts by default.

---

# 6. Countplot

A countplot displays the frequency/count of each category.

It simply counts:

* how many times each category appears

### Main Purpose

* Analyze categorical frequency distribution

### Example Use Cases

* Male vs female count
* Product category frequencies
* Number of students in each grade

### Advantages

* Very useful for categorical EDA
* Helps identify class imbalance

---

# Summary Table

| Plot       | Main Purpose                            |
| ---------- | --------------------------------------- |
| Boxplot    | Statistical summary + outlier detection |
| Violinplot | Distribution + density visualization    |
| Stripplot  | Show individual data points             |
| Swarmplot  | Non-overlapping individual points       |
| Barplot    | Compare aggregated numerical values     |
| Countplot  | Show category frequencies/counts        |
