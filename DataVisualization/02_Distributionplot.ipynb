## 1. Histplot / Distplot

### Histplot

A histogram plot used to visualize the distribution of numerical data.

It divides data into intervals called **bins** and shows:

* frequency/count of values
* spread of data
* skewness
* outliers

#### Used For

* Understanding data distribution
* Detecting concentration of values
* Checking normal distribution

---

### Distplot

A distribution plot (older Seaborn plot, now mostly replaced).

It combines:

* histogram
* KDE curve (density curve)

to show both:

* frequency
* probability density

#### Used For

* Understanding shape of data
* Statistical analysis
* Distribution comparison

---

## 2. Jointplot

A jointplot shows:

* relationship between two variables
* along with their individual distributions

It combines:

* scatter plot (center)
* histograms/distributions (sides)

#### Used For

* Correlation analysis
* Relationship detection
* Bivariate analysis

#### Helps Identify

* positive/negative correlation
* clusters
* outliers

---

## 3. Pairplot

A pairplot creates pairwise relationships between multiple numerical columns in a dataset.

It automatically generates:

* scatter plots between every pair of variables
* histograms/distributions on diagonal

#### Used For

* Exploratory Data Analysis (EDA)
* Finding patterns
* Detecting correlations
* Feature relationship analysis

#### Very Useful In

* Data Science
* Machine Learning preprocessing

---

## 4. Rugplot

A rugplot displays small tick marks for every individual data point on an axis.

Each line/tick represents:

* one observation/value

#### Used For

* Seeing exact distribution of points
* Understanding data density
* Supporting KDE/distribution plots

#### Characteristics

* Minimalistic plot
* Often combined with histplot/distplot
* Good for small datasets
