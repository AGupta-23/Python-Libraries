```python
import numpy as np
import pandas as pd
```

## Finding missing Values - Identifying

- Col wise he hota hai!

```python
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [np.nan, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
df = pd.DataFrame(data)
df
#np.nan are null values
```

```python
df.isna()
#if null value then True
```

```python
df.isna().sum()
# have to find out which row have max number of null values
```

```python
df.isna().any()
# cols me null value hai ki nai?
```

## Removing Missing Values

works on basis of ROWS

```python
df
```

```python
df.dropna()
# all rows had null values - so all rows dropped
```

```python
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [1, 2, 3, 4, 5],          #changed 1st value -- row non null
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
df = pd.DataFrame(data)
df
```

```python
df.dropna()
```

```python
df.dropna(thresh=3)
# atleast 3 non null values in row
```

```python
df.dropna(thresh=4)
```

## Filling the missing data

- Col wise hota hai!

```python
df.fillna(0)

#non null values become 0
```

```python
df 
# same only until we say - *inplace=True*
```

```python
# Fill missing values with diff value for each col

#created a dict
values = {'A': 0, 'B': 100, 'C': 300, 'D': 400}
df.fillna(value=values)
```

```python
df
```

```python
# To fill mean

df.fillna(df.mean())
# which was 3.0 whole dataframe ka
```
