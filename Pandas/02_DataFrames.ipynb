## Creating a DataFrame
Can use list, dictionary, arrays etc

```python
data = { #dict 
    'Name': ['John', 'Anna', 'Peter', 'Linda'],
    'Age': [28, 34, 29, 42],
    'City': ['New York', 'Paris', 'Berlin', 'London'],
    'Salary': [65000, 70000, 62000, 85000]
}
df = pd.DataFrame(data)
```

```python
df  #multiple series combined
```

```python
# Also create data frame using list! --> does not have keys

data_list = [
    ['John', 28, 'New York', 65000],
    ['Anna', 34, 'Paris', 70000],
    ['Peter', 29, 'Berlin', 62000],
    ['Linda', 42, 'London', 85000]
]
df2 = pd.DataFrame(data_list)
# col names - headers default goes - 0,1,2,3
# row names - default always 1,2,3,4,..

columns = ["Name","Age","City","Salary"]
df2 = pd.DataFrame(data_list,columns =columns) # manually set
df2
```

```python
# thats why inplace= True used1
a=df2.drop("Designation", axis=1, inplace=True)
a
```
