## Pandas Series

A series is a one-dimentional labeled array capable of holding **ANY DATA TYPE**. The axis labels are collectively called the index. 

- arr=[1,2,3,4,5] --> horizontal
- series --> Vertical 

## Creating series

There are multiple ways to create pandas Series

```python
#!pip install pandas
```

```python
import numpy as np
import pandas as pd
```

```python
labels=["a","b","c"]  # act as index!
mylist=[10,20,30]
arr = np.array([10,20,30])
d ={1:10,2:20,3:30}
```

```python
pd.Series(mylist) #will print vertical
```

```python
#set custom labels

pd.Series(mylist,index=labels)
```

```python
pd.Series(arr,index=labels)  #will give error in matrix as series is 1D only
```

```python
# pd.Series(d, index=labels)
# a   NaN
# b   NaN
# c   NaN
# dtype: float64

pd.Series(d)
```
