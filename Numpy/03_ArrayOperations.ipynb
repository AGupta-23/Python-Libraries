#### Arithmetic Operations in NumPy Arrays

Arithmetic operations happen between elements at the **same positions**.

Example:

```python
import numpy as np

a = np.array([1,2,3])
b = np.array([4,5,6])

print(a + b)
```

Output:

```python
[5 7 9]
```

Because:

```python
1+4
2+5
3+6
```

---

#### Broadcasting in NumPy

Broadcasting means NumPy automatically expands a smaller value/array to match the shape of a bigger array.

Example:

```python
a = np.array([1,2,3])

print(a + 10)
```

Output:

```python
[11 12 13]
```

Here NumPy “broadcasts” `10` like this internally:

```python
[10 10 10]
```

Then performs arithmetic.

---

#### Why Broadcasting is Different

Normal arithmetic operations:

* Need arrays of same size
* Element-by-element operation

Broadcasting:

* Allows different shapes/sizes
* NumPy automatically adjusts smaller data

Example:

```python
a = np.array([1,2,3])
b = np.array([[10],[20],[30]])

print(a + b)
```

Output:

```python
[[11 12 13]
 [21 22 23]
 [31 32 33]]
```

NumPy automatically expanded shapes to make operation possible.

---

#### Short Definition

Broadcasting = Automatic expansion of smaller arrays/scalars so arithmetic operations can happen between different-sized arrays.


## Deep and Shallow Copy

works in list as well

````python
a= np.arange(1,21)
a
#prints vector
````

````python
slice= a[:5]
slice
````

````python
#broadcasting in slice
slice=slice*10
slice
````

````python
a
#but a does not change -original is same! --> SHALLOW COPY
#It is not such in list
````

````python
a=a[4:6]*10
a
#DEEP COPY - CHANGES ORIGINAL ARRAY
````

````python
b=a  #copied -- so changes in b will change a as well --> DEEP COPY
b[0]=999
b
````

````python
a
````

````python
b=a.copy()
b[1]=888
print(b)
print(a)  #does not change as copy of a passed
````

## Advance Array Manipulation 

### 1. Stacking of arrays

````python
#You cannot stack 2 matrix 
#only stack vectors - 1D array

a=np.array([1,2,3,4])
b=np.array([5,6,7,8])
````

````python
#Vertical stacking
np.vstack((a,b))   #-->takes tuple arguments
````

````python
#Horizontal Stack
np.hstack((a,b))
````

````python
#Column Stack
np.column_stack((a,b))
````
