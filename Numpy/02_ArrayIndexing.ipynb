## Indexing and Slicing of Vectors

```python
import numpy as np
```

```python
arr=np.arange(11,21)
arr
```

#### Indexing

```python
int(arr[5]) #-->16 python integer

#while normally give numpy integer
# NumPy uses its own datatypes because they are:
# Faster
# Memory efficient
# Better for scientific computing
```

#### Slicing

```python
arr[1:5] # [start:end]
arr[:5] # [from start to end point given]
arr[3::]  #from 3 till end index
arr[3::2] #from 3 skip 2 index and give output
```

## Numpy Indexing and Slicing of Matrix 

```python
arr=np.arange(1,31).reshape(6,5)
arr
```

```python
arr[0]
```

```python
arr[2]
```

```python
arr[4,3]
#element at 5th row and 4rd column
#indexing starts from 0
```

```python
arr[0:2,1:3]

#array([[2, 3],
    #    [7, 8]])
```

```python
arr[3:,3:]
```

```python
#selecting only col
#becomes a vector
arr[:,2]
```

## Boolean Indexing

```python
arr=np.arange(11,21)
arr
```

```python
#want odd only n remove even

Bool_index = arr % 2 !=0
Bool_index
```

```python
arr=arr[Bool_index]
arr
```

```python
arr=np.arange(11,21)
arr
```

```python
#want even only
Bool_index = arr % 2 ==0
arr=arr[Bool_index]
arr
```
