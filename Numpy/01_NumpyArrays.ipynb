## Array Generation Function

```python
arr=np.arange(1,11) #works like range function of loop
arr
```

```python
arr=np.zeros(6)
arr
```

```python
arr=np.zeros((6,6))
arr
#horizontal are rows
#vertical are columns
```

```python
arr=np.ones(7)
arr
```

```python
arr=np.ones((4,7))
arr
```

```python
#linspace or linear spacing vs range func
#Return evenly spaced numbers over a specified interval.
arr = np.linspace(1,5,3)
#(linspace(start,stop,number of elements we want)) -- and it will provide equal spaces between
arr
```

## Random Generation Function

Random library - random integers 
add with numpy library - random array 
Also downloaded with numpy automatically
Random library - inbuilt

```python
np.random.rand(5)
#np,random--libraries
#rand-->function
#produces vector --> all values between 0 and 1 --> concept of normalization!!
```

```python
np.random.randn(10)
#randn -> produces -3 to +3 --> concept of standardization
```

```python
np.random.randint(6)
#gives between (0,6)
#output is a single digit
```

```python
np.random.randint(10,20,10)
#Produces 10 element array between 10,20
```

## Array Methods

```python
min=arr.min()
print(min)
max=arr.max()
print(max)
```

```python
summation=arr.sum()
print(summation)
```

```python
np.sum(arr,axis=0)
#gives summation of columns = axis =0
```

```python
np.sum(arr,axis=1)
#gives row summation
```

```python
print(arr.mean())
```

```python
arr.std() #gives standard deviation
```

```python
arr.argmax() #gives max number ka index
```

```python
arr.argmin() #gives min ka index
```
